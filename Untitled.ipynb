## Artifact Glossary

| Artifact | Format | Where Stored |
|---|---|---|
| `dataset_hash` | SHA-256 string | MLflow params + DVC `.dvc` file |
| `run_id` | UUID (MLflow) | MLflow Tracking Server |
| `training_config.yaml` | YAML | Git commit (pinned to run_id) |
| `eval_report.json` | JSON | MLflow artifacts |
| `model URI` | `models:/eta/<stage>#<version>` | MLflow Model Registry |
| `deployed_version` | Git tag `eta-v{N}-{canary\|prod}` | Git + K8s label |
| `drift_signal.json` | JSON (PSI per feature, rolling MAE) | Monitoring store (EvidentlyAI export) |

## Transition Legend

- ** AUTO** — triggered by CI/CD pipeline or monitoring hook, no human required  
- ** MANUAL** — requires explicit approval in the registry UI or PR review

# ADR 0001: Reproducibility Strategy for NorthStar Models

**Status:** Accepted  
**Date:** 2025-06-05  
**Deciders:** ML Lead, Platform Lead

---

## Context

NorthStar's ML team currently stores model artifacts in S3 with ad-hoc names
like `eta_v2_FINAL.onnx`. There is no dataset versioning, no experiment
tracker, and no policy for pinning training environments. Reproducing any past
result requires tribal knowledge of which S3 folder, which laptop, and which
undocumented library versions were involved — a situation that has already
blocked two post-incident root-cause analyses.

---

## Decision

We adopt a four-layer reproducibility contract. Every training run that
produces a registered artifact must satisfy all four layers; CI gates enforce
this at promotion time.

### Layer 1 — Environment

All training jobs run inside a Docker image built from a `Dockerfile` pinned to
a specific base image digest (e.g. `python:3.11.9-slim@sha256:7f3a…`).
Python dependencies are declared in `requirements.txt` with exact versions
(`torch==2.3.0`, `scikit-learn==1.4.2`, etc.) and a `requirements.lock` file
generated by `pip-compile` is committed to Git.  The image digest
(`base_image_sha`) is a required lineage field in the model registry; a run
whose image cannot be identified by digest is rejected at staging promotion.

### Layer 2 — Data

All datasets are versioned with **DVC** backed by the team's S3 bucket
(`s3://northstar-dvc-store`).  Every training job receives a DVC ref
(e.g. `data/train/eta_2024q4.dvc`) and records the SHA-256 content hash of
the resolved snapshot as `dataset_hash` in MLflow params.  The frozen
evaluation sets (`eta_frozen_v7.dvc`, `dispatcher_frozen_v3.dvc`) are
declared in `model-registry.yaml` and are never updated mid-lifecycle; a new
eval set requires a new ADR revision and bumps the `eval_set_ref` field.

### Layer 3 — Code

All training code lives in the monorepo under `src/models/`.  The exact Git
commit SHA (`git_sha`) is logged as an MLflow param at run start via a
pre-training hook in `train.py`.  Feature-engineering logic is treated as
code (not notebook cells): all transformations are in importable Python
modules, covered by unit tests, and diffable in PRs.  Notebooks are
permitted for exploration only and must never be the source of a registered
model.

### Layer 4 — Randomness

Every training script accepts a `--seed` argument (default `42`) which is
passed to `random.seed()`, `numpy.random.seed()`, and
`torch.manual_seed()` / `torch.cuda.manual_seed_all()`.  The seed value is
logged as an MLflow param.  PyTorch's `torch.use_deterministic_algorithms(True)`
is enabled for training runs that produce registered artifacts; the flag may be
disabled for quick exploratory runs but the resulting models cannot be promoted
to staging.

---

## Alternatives Rejected

- **S3 timestamps as dataset versioning** — Object last-modified timestamps are
  mutable (overwritten by re-upload, altered by S3 lifecycle rules) and
  provide no content integrity guarantee. Two files can share a timestamp or
  differ in content with the same timestamp. Rejected: fails the integrity
  requirement.

- **Conda environments checked into Git** — `environment.yml` files list
  package names with optional version constraints but do not lock transitive
  dependencies. Solving the environment on a different platform produces
  different resolved versions. Rejected: reproduces intent, not the exact
  environment. We use Docker + `pip-compile` instead.

- **MLflow autolog only, no explicit seed policy** — MLflow autolog captures
  framework hyperparameters but does not enforce deterministic algorithm
  selection or capture the global random seed across all libraries. Silent
  non-determinism from CUDA kernels has caused 2–3% metric variance between
  runs. Rejected: variance contaminates gate decisions.

---

## Consequences

- Every engineer must install DVC and configure the S3 remote (`dvc remote add`)
  before running any training job. A one-time setup script (`scripts/setup_dvc.sh`)
  is provided and documented in the contributing guide.
- CI pipelines must have access to the S3 DVC store and the Docker registry;
  adding a new training environment requires a platform ticket, not a one-liner.
- `torch.use_deterministic_algorithms(True)` may reject certain CUDA operations
  that lack deterministic implementations; the team must replace those ops or
  document the non-deterministic path explicitly when encountered.

---

## Revisit If

- The team moves to a data-warehouse-first architecture (e.g. Snowflake feature
  store) where dataset versions are expressed as query snapshots rather than
  file hashes — at that point DVC file hashing no longer captures the full
  lineage and this ADR must be updated to version SQL queries and table
  snapshots instead.
